# Isolated offline scoring
Use a dedicated Linux evaluator with a reviewed checkout and pinned evaluator venv. Do not mount Drive or copy credentials into this environment. Bubblewrap must pass capability probes; notebook separation alone is not isolation.

```python
from pathlib import Path
import subprocess, sys, os
CHECKOUT = Path('/content/Pact')  # separately pinned reviewed checkout
EVALUATOR_PYTHON = Path('/content/pact-evaluator/bin/python')
JOBS = Path('/content/scoring_jobs.json')  # private, checksum-verified transfer
OUTPUT = Path('/content/scoring_results.json')
EXECUTE = False
os.environ['PYTHONPATH'] = str(CHECKOUT/'src')
```

```python
subprocess.run([sys.executable,'-m','pact','benchmarks','code-probe','--evaluator-python',str(EVALUATOR_PYTHON)],cwd=CHECKOUT,check=True)
```

```python
if EXECUTE:
    subprocess.run([sys.executable,'-m','pact','benchmarks','code-run','--evaluator-python',str(EVALUATOR_PYTHON),'--jobs',str(JOBS),'--output',str(OUTPUT)],cwd=CHECKOUT,check=True)
else:
    print('Probe only. No generated program executed.')
```
